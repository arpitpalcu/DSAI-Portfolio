## Daily Demand Forecasting Dataset

```python
# Aggregate transaction demand to daily demand

daily_forecast = (df.groupby("Date")
      .agg(Demand=("Order_Demand", "sum"), Orders=("Order_Demand", "count"),
           Promo_Rate=("Promo", "mean"), SchoolHoliday_Rate=("SchoolHoliday", "mean"))
    .reset_index())

daily_forecast = daily_forecast.sort_values("Date").reset_index(drop=True)

display(daily_forecast.head())
print("Daily dataset shape:", daily_forecast.shape)
```

```python
# Create calendar features

daily_forecast["Year"] = daily_forecast["Date"].dt.year
daily_forecast["Month"] = daily_forecast["Date"].dt.month
daily_forecast["DayOfWeek"] = daily_forecast["Date"].dt.dayofweek
daily_forecast["WeekOfYear"] = (daily_forecast["Date"].dt.isocalendar().week.astype(int))

display(daily_forecast.head())
```

```python
# Create lag features

daily_forecast["Lag_1"] = (daily_forecast["Demand"].shift(1))

daily_forecast["Lag_7"] = (daily_forecast["Demand"].shift(7))

daily_forecast["Lag_14"] = (daily_forecast["Demand"].shift(14))

daily_forecast["Lag_28"] = (daily_forecast["Demand"].shift(28))
```

```python
# Create rolling demand features
# Shift first to prevent future-data leakage

daily_forecast["Rolling_Mean_7"] = (daily_forecast["Demand"].shift(1).rolling(7).mean())

daily_forecast["Rolling_Mean_28"] = (daily_forecast["Demand"].shift(1).rolling(28).mean())

daily_forecast["Rolling_Std_7"] = (daily_forecast["Demand"].shift(1).rolling(7).std())
```

```python
# Remove rows created by lag/rolling calculations

forecast_data = daily_forecast.dropna().reset_index(drop=True)

print("Original daily rows:", len(daily_forecast))
print("Forecasting rows:", len(forecast_data))
print("Missing values:", forecast_data.isnull().sum().sum())

display(forecast_data.head())
```

```python
# Correlation with demand

forecast_correlations = (forecast_data
    [["Demand", "Lag_1", "Lag_7", "Lag_14", "Lag_28", "Rolling_Mean_7", "Rolling_Mean_28", 
      "Rolling_Std_7", "Promo_Rate", "SchoolHoliday_Rate"]].corr()["Demand"]
    .sort_values(ascending=False))

display(forecast_correlations)
```

```python
# Final forecasting dataset check

print("=" * 60)
print("FORECASTING DATASET QA")
print("=" * 60)

print("Rows:", forecast_data.shape[0])
print("Columns:", forecast_data.shape[1])

print("Date range:", forecast_data["Date"].min(), "to", forecast_data["Date"].max())

print("Missing values:", forecast_data.isnull().sum().sum())
```

```python
# Time-based train/test split

feature_columns = [
    "Lag_1",
    "Lag_7",
    "Lag_14",
    "Lag_28",
    "Rolling_Mean_7",
    "Rolling_Mean_28",
    "Rolling_Std_7",
    "Promo_Rate",
    "SchoolHoliday_Rate",
    "Month",
    "DayOfWeek",
    "WeekOfYear"
]

target_column = "Demand"

split_index = int(len(forecast_data) * 0.8)

train_data = forecast_data.iloc[:split_index].copy()
test_data = forecast_data.iloc[split_index:].copy()

X_train = train_data[feature_columns]
y_train = train_data[target_column]

X_test = test_data[feature_columns]
y_test = test_data[target_column]

print("Training rows:", len(train_data))
print("Testing rows:", len(test_data))
```

### Train Random Forest

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(n_estimators=150, random_state=42, n_jobs=-1)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)

print("Random Forest trained successfully.")
```

### Train Gradient Boosting

```python
from sklearn.ensemble import HistGradientBoostingRegressor

gb_model = HistGradientBoostingRegressor(max_iter=200, learning_rate=0.05, random_state=42)

gb_model.fit(X_train, y_train)

gb_predictions = gb_model.predict(X_test)

print("Gradient Boosting trained successfully.")
```

### Compare models

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

def calculate_metrics(actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    return mae, rmse


rf_mae, rf_rmse = calculate_metrics(y_test, rf_predictions)

gb_mae, gb_rmse = calculate_metrics(y_test, gb_predictions)

model_comparison = pd.DataFrame({"Model": ["Random Forest", "Gradient Boosting"],
                                 "MAE": [rf_mae, gb_mae], "RMSE": [rf_rmse, gb_rmse]}).round(2)

display(model_comparison)
```

### Actual vs predicted

```python
test_results = test_data[["Date", "Demand"]].copy()

test_results["Random_Forest"] = rf_predictions
test_results["Gradient_Boosting"] = gb_predictions

display(test_results.head())
```

### Forecast visualization

```python
plt.figure(figsize=(14, 6))

plt.plot(test_results["Date"], test_results["Demand"], label="Actual")

plt.plot(test_results["Date"], test_results["Random_Forest"], label="Random Forest")

plt.plot(test_results["Date"], test_results["Gradient_Boosting"], label="Gradient Boosting")

plt.title("Actual vs Predicted Retail Demand")
plt.xlabel("Date")
plt.ylabel("Demand")
plt.legend()

plt.tight_layout()

plt.savefig(FIGURE_DIR / "04_forecast_vs_actual.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
feature_importance = pd.DataFrame({
    "Feature": feature_columns, "Importance": rf_model.feature_importances_}
                                 ).sort_values("Importance",ascending=False)

display(feature_importance)
```

### Inventory intelligence

```python
inventory_intelligence = test_results.copy()

inventory_intelligence["Forecast_Demand"] = (inventory_intelligence["Gradient_Boosting"])

inventory_intelligence["Forecast_Error"] = (inventory_intelligence["Demand"]
                                            - inventory_intelligence["Forecast_Demand"])

inventory_intelligence["Demand_Risk"] = np.where(inventory_intelligence["Forecast_Demand"]
    > inventory_intelligence["Forecast_Demand"].quantile(0.75), "High", np.where(
    inventory_intelligence["Forecast_Demand"]
    < inventory_intelligence["Forecast_Demand"].quantile(0.25), "Low", "Medium"))

display(inventory_intelligence.head(10))
```

### Final summary

```python
best_model = (model_comparison.sort_values("RMSE").iloc[0]["Model"])

print("=" * 70)
print("RETAIL DEMAND FORECASTING — FINAL RESULTS")
print("=" * 70)

print(f"\nTraining observations: {len(train_data):,}")
print(f"Testing observations: {len(test_data):,}")

print("\nMODEL PERFORMANCE")
display(model_comparison)

print(f"Best model by RMSE: {best_model}")

print("\nTOP PREDICTIVE FEATURES")
display(feature_importance.head(10))

print("\nINVENTORY RISK DISTRIBUTION")
print(
    inventory_intelligence["Demand_Risk"]
    .value_counts()
)

print("\nFINAL QA COMPLETE")
```

## Conclusion

This project transformed retail transaction data into a demand
forecasting and inventory intelligence framework.

The analysis examined demand trends, seasonality, warehouse patterns,
product categories, promotion activity, and historical demand behavior.

Lag-based and rolling demand features were used to build machine
learning forecasting models. Random Forest and Gradient Boosting were
evaluated using MAE and RMSE, with the better-performing model selected
based on test-set RMSE.

The final inventory intelligence layer classifies forecast demand into
relative demand-risk levels to support inventory planning.

These results are analytical estimates based on historical observations
and should support, rather than replace, operational inventory decisions.
